```python
#RANDOM TENSORS
#these are used to update weights and biases?

#Std tandom tensor
Random_tensor = torch.rand(3,4)
print(Random_tensor)
print(Random_tensor.ndim)
print(Random_tensor.shape)

#Using random generator
g = torch.Generator().manual_seed(42)
c = torch.rand(2, 3, generator=g)
print("c =", c)

# Output to an existing tensor
out_tensor = torch.empty(2, 3)
torch.rand(2, 3, out=out_tensor)
print("out_tensor =", out_tensor)

# Enable autograd
d = torch.rand(2, 3, requires_grad=True)
print("d =", d)
```

```python
#Creating a random tensor with similar shape to an image tensor
random_image_size_tensor = torch.rand(size=(224, 224, 3)) #height, width, color channel (r,g,b)
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

```python
# Zeros and ones
zeros = torch.zeros(size=(3,4))
ones = torch.ones(size=(3,4))
print(zeros)
print(ones)
print(zeros*Random_tensor)
```

##TENSOR DataType

```python
#Float 32 tensor is the default datatype
float_32_Tensor = torch.tensor([3.0,6.0,9.0], dtype=None, #default is f32
                                              device=None,
                                              requires_grad=False)
float_32_Tensor
```

```python
float_32_Tensor.dtype
```

### TENSOR datatypes
**Note:** Tensor datatypes is one of the 3 big errors with Pytorch and deep learning:
1. Tensors not right datatype (randomly changing precision can cause erors) -> use 'tensor.dtype'
2. Tensors not right shape (for tensor multiplication) -> use 'tensor.shape'
3. Tensors not on the right device (either CPU or GPU) -> use 'tensor.device'

```python
#converting float 32 to float 16
float_16_tensor = float_32_Tensor.type(torch.float16)
float_16_tensor
```

```python
#multiplying f32 x f16
ans = float_16_tensor*float_32_Tensor
ans, ans.dtype
```

```python
#int32xfloat32
int_32_tensor = torch.tensor([3,6,9], dtype=torch.int32)
ans = int_32_tensor*float_32_Tensor
ans, ans.dtype # converts to float 32.
```
