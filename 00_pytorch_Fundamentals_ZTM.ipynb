### TENSOR datatypes
**Note:** Tensor datatypes is one of the 3 big errors with Pytorch and deep learning:
1. Tensors not right datatype (randomly changing precision can cause erors) -> use 'tensor.dtype'
2. Tensors not right shape (for tensor multiplication) -> use 'tensor.shape'
3. Tensors not on the right device (either CPU or GPU) -> use 'tensor.device'

````python
#converting float 32 to float 16
float_16_tensor = float_32_Tensor.type(torch.float16)
float_16_tensor
````

### MAnipulating tensors (tensor operations)

Tensor operations include:
* Addition
* Substraction
* Multiplication (element wise)
* Division
* Matrix multiplication
* Dot product


````python
# Create a tensor
tensor = torch.tensor([1,2,3])
print(tensor + 10) #addition
print(tensor * 10) #multiplication
print(tensor - 10) #subtraction
````

````python
#Trying out inbuilt functions of pytorch.
print(torch.add(tensor, 10)) #addition
print(torch.mul(tensor, 10)) #multiplication
print(torch.sub(tensor, 10)) #subtraction
````

````python
# MAtrix multiplication
tensor * tensor
torch.matmul(tensor, tensor)
````

````python
#using @ for Matmul
print(tensor @ tensor)
````

````python
#Lets us time matrix multiplication using For look and internal method.
#internal methods are faster cause the multiplication would require us to retrieve things first.
# dont use print. The print makes it take double the time.
%%time
value = 0
for i in tensor:
  value += i*i
value
````

````python
#Note when using it dont use print. The print also takes up time.
%%time
torch.matmul(tensor, tensor)
````

## Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - reshapes an input tensor to a defined shape
* View - return a view of an input tensor of certain shape but keeps the memory as the original tensor
* Stacking - combines multiple tensors of different types.
            vstack for stacking vertically and hstack for stacking size by side. Think concatenate.
* Squeeze - removes all `1` dimensions from a tensor
* Unsqueeze - add a `1` dimension to a target tensor.cdouble
* Permute - Return a view of the input with dimensions permuted (swapped) in a certain way

````python
import torch
x = torch.arange(0,10,1)
x, x.shape
````

````python
# Add an extra dimension
x_reshaped = x.reshape(1,10)
x_reshaped, x_reshaped.shape #Notice the extra bracket
````

````python
# Change view
z = x.view(2,5)
z, z.shape
````

````python
#If i change z, x changes. View of a tensor shares the same memory as the original tensor.
#Even though the shape does change, the data stored changes.
z[:, 0] = 5
z, x
````

````python
#Stacking tensors
x_stacked = torch.stack([x,x,x,x], dim=0)
print(x_stacked)
x_stacked = torch.stack([x,x,x,x], dim=1)
print(x_stacked)
````

````python
#Squeeze and unsqueeze
print(f"previous version")
print(x_reshaped)
print(x_reshaped.shape)
print(f"="*30)
print(f"Squeezed version")
x_squeezed = x_reshaped.squeeze()
print(x_squeezed)
print(x_squeezed.shape) # see 1 bracket less.
````

````python
# unsqueeze
print(f"Squeezed version of out tensor")
print(x_squeezed)
print(x_squeezed.shape)
print(f"="*30)
print(f"Unsqueezed version")
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(x_unsqueezed)
print(x_unsqueezed.shape) # adde extra brackets [a] becomes [[a]]
print(f"="*30)
print(f"Unsqueezed version for dim = 1")
x_unsqueezed = x_squeezed.unsqueeze(dim=1)
print(x_unsqueezed)
print(x_unsqueezed.shape) # adde extra brackets to the 1st dimension. [a] becomes [ax1]
````

````python
#permute
x_original = torch.rand(size=(224,224,3)) # [h, w, c_c]
x_permuted = x_original.permute(2,0,1) # rearrange to [c_c, h, w]
x_permuted.shape
````

````python
#BEcause permutation is a view, changing the view, changes the original.
#Moreover, changing the original changes the view.
x_original[0,0,0] = 728218
x_permuted[0,0,0], x_original[0,0,0]
````

## Pytorch Tensors and numpy

Pytorch has funcitonality to interact wiht it.

* Data in numpy, want in pytorch tensor ->  `torch.from_numpy(ndarray)`
* Pytorch tensor to numpy -> `torch.Tensor.numpy()`

##IMP
* default dtype in numpy if f64, contrast with pytorch which is f32.
* Moreover, they don't share memory. Thus changing one does not affect the other.

````python
array = np.arange(1.0,8.0)
tensor = torch.from_numpy(array)
array, tensor # numpy's default dtype is f64
````

````python
tensor =  torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
````

## Moving tensors form CPU to GPU

````python
import torch
device = "cuda" if torch.cuda.is_available() else "cpu"
print(device)
tensor = torch.tensor([1, 2, 3])

#Tensor not on GPU
print(tensor, tensor.device) # by default tensors created on cpu
````

````python
# move tensor to GPU
tensor_on_gpu = tensor.to(device)
tensor_on_gpu # the index of the gpu is also returned
````

NUMPY can only be applied to tensors on **CPU** not on **GPU**

````python
# move tensors to CPU
tensor_back_on_cpu = tensor_on_gpu.cpu().numpy()
tensor_back_on_cpu
````
